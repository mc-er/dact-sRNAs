## GO enrichment analysis for DT genesusing topGO

```r
library("tidyverse")
library("dplyr")
library("topGO")
```

Set the working directory and source script with functions

```r
setwd("/home/mier0006/Documents/projects/phd_dact/smRNA_dact/")
source("dact-sRNAs/scripts/used_functions.R")
```

Set some variables

```r
fp_GOdb <- "dact-sRNAs/data/Dinc_blastOrmRNA-MB1-CDS_geneID-2go_gene-ID.map.gz"
fp_dt_polyploids <- "05_topGO/gene-IDs/TM_20to24_all.txt"
fp_dt_maj <- "05_topGO/gene-IDs/transgressive_maj_all.txt"
fp_dt_tra <- "05_topGO/gene-IDs/transgressive_tra_all.txt"

GO_category <- "BP"
GO_algorithm <- "weight01"
GO_statistic <- "fisher"
```

topGO enrichment analysis for DT genes between polyploids (i.e. *D. majalis* and *D. traunsteineri*).

```r
enrichment_polyploids <- enriched_GO(fp_GOdb, fp_dt_polyploids, GO_category, GO_algorithm, GO_statistic)

write.table(enrichment_polyploids, "05_topGO/tra-maj_BP_20to24nt.txt",
            sep="\t", col.names=T, row.names=F, quote=F)

enrichment_polyploids %>%
    filter(adj_pval < 0.05) %>%
    write.table("05_topGO/tra-maj_BP_20to24nt_sig-0.05.txt",
            sep="\t", col.names=T, row.names=F, quote=F)
```

topGO enrichment analysis for transgressive DT genes between *D. majalis* and either diploid (i.e. *D. fuchsii* and *D. incarnata*).

```r
enrichment_maj <- enriched_GO(fp_GOdb, fp_dt_maj, GO_category, GO_algorithm, GO_statistic)

write.table(enrichment_maj, "05_topGO/transgressive_maj.txt",
            sep="\t", col.names=T, row.names=F, quote=F)

enrichment_maj %>%
    filter(adj_pval < 0.05) %>%
    write.table("05_topGO/transgressive_maj_sig-0.05.txt",
            sep="\t", col.names=T, row.names=F, quote=F)
```

topGO enrichment analysis for transgressive DT genes between *D. traunsteineri* and either diploid (i.e. *D. fuchsii* and *D. incarnata*).

```r
enrichment_tra <- enriched_GO(fp_GOdb, fp_dt_tra, GO_category, GO_algorithm, GO_statistic)

write.table(enrichment_tra, "05_topGO/transgressive_tra.txt",
            sep="\t", col.names=T, row.names=F, quote=F)

enrichment_tra %>%
    filter(adj_pval < 0.05) %>%
    write.table("05_topGO/transgressive_tra_sig-0.05.txt",
            sep="\t", col.names=T, row.names=F, quote=F)
```
